## Streaming and Batch

## Streaming

```python
model.invoke("write me a 200 word essay on the history of the internet")
```

```python
model.stream("write me a 200 word essay on the history of the internet")
```

```python
for chunk in model.stream("write me a 200 word essay on the history of the internet"):
    print(chunk.text, end="|", flush=True)
```

```python
for chunk in model.stream("why do parrots have colorful feathers?"):
    print(chunk.text, end="|", flush=True)
```

## Batch

Batching a collection of independent requests to a model can significantly improve performance and reduce the costs, as the processing can be done in parallel

```python
responses = model.batch([
    "why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is the capital of France?"
])

for response in responses:
    print(response.content)
```

```python
responses = model.batch([
    "why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is the capital of France?"
], config={"max_concurrency": 5})

for response in responses:
    print(response.content)
```
